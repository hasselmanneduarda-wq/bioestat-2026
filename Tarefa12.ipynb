#FRASES:
##Bernoulli é adequada quando será analisado um evento com somente dois resultados possíveis (sucesso e fracasso).

##Binomial é adequada quando o número de sucessos em uma quantidade fixa de ensaios independentes é contado e há a observação de uma probabilidade constante.

##Poisson é adequada quando Poisson é adequada quando se utiliza o número de eventos que ocorrem em um intervalo contínuo de tempo ou espaço, conhecendo apenas a taxa média de ocorrência.


```python
from scipy.stats import binom, poisson

# Exercício 1. Qual modelo usar?

# B) 20 sementes, p = 0.8, exatamente 15 germinam
prob_1b = binom.pmf(15, 20, 0.8)
print(f"1B - P(X=15): {prob_1b:.2f}")

# C) Média de 4 ninhos por hectare, exatamente 6 ninhos
prob_1c = poisson.pmf(6, 4)
print(f"1C - P(X=6): {prob_1c:.2f}")


# Exercício 2. População de abelhas

n = 10
p = 0.30

# A) Distribuição:
# X ~ Binomial(10, 0.30)

# B) Exatamente 3 abelhas
prob_2b = binom.pmf(3, n, p)
print(f"2B - P(X=3): {prob_2b:.2f}")

# C) Pelo menos 1 abelha
prob_2c = 1 - binom.pmf(0, n, p)
print(f"2C - P(X>=1): {prob_2c:.2f}")


# Exercício 3. Observação ecológica

lamb = 2

# B) Exatamente 3 encontros
prob_3b = poisson.pmf(3, lamb)
print(f"3B - P(X=3): {prob_3b:.2f}")

# C) Nenhum encontro
prob_3c = poisson.pmf(0, lamb)
print(f"3C - P(X=0): {prob_3c :.2f}")

# D) Mais de 2 encontros
prob_3d = 1 - poisson.cdf(2, lamb)
print(f"3D - P(X>2): {prob_3d:.2f}")
```
