```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_excel('/content/tarefa03.xlsx')
display(df)
```

```python
# 1. Frequência absoluta da variável nominal: Você lava as mãos antes da refeição?
freq_lava = df["Você lava as mãos antes da refeição?"].value_counts()
freq_lava
```

```python
# 2. Frequência absoluta da variável discreta: Quantas vezes lava por dia?
freq_vezes = df["Quantas vezes lava por dia?"].value_counts().sort_index()
freq_vezes
```

```python
# 3. Frequência absoluta da variável discreta: Conhecimento em microbiologia (0 – 10)
freq_conhec = df["Conhecimento em microbiologia (0 – 10)"].value_counts().sort_index()
freq_conhec
```

```python
# 4. Frequência absoluta da variável discreta: Faz uso de álcool em gel?
freq_alcool = df["Faz uso de álcool em gel?"].value_counts()
freq_alcool
```

```python
freq_a = df["Faz uso de álcool em gel?"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    freq_a,
    labels=freq_a.index,
    autopct="%.1f%%",
    colors=["#F8BBD0", "#CE93D8", "#90CAF9"]
)

plt.title("Frequência de utilização de álcool em gel (%)")
plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.histplot(
    data=df,
    x="Conhecimento em microbiologia (0 – 10)",
    discrete=True
)

plt.title("Conhecimento em microbiologia (0 – 10)")
plt.xlabel("Nota atribuída")
plt.ylabel("Frequência")

plt.xticks(sorted(df["Conhecimento em microbiologia (0 – 10)"].unique()))

plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.scatterplot(
    data=df,
    x="Quantas vezes lava por dia?",
    y="Conhecimento em microbiologia (0 – 10)"
)

plt.title("Quantas vezes lava por dia × Conhecimento em microbiologia")
plt.xlabel("Quantidade de vezes que lava as mãos por dia")
plt.ylabel("Conhecimento em microbiologia (0 – 10)")

plt.xticks(sorted(df["Quantas vezes lava por dia?"].unique()))
plt.yticks(range(0, 11))

plt.show()
```

#Relatório

## Através da análise dos dados obtidos, com 19 estudantes, observamos uma relação positiva entre o conhecimento sobre microbiologia e a frequência de lavagem das mãos por dia. Dessa forma, os estudantes que afirmaram maior conhecimento em microbiologia também relataram uma maior frequência de lavagem das mãos. No gráfico de dispersão, observa-se uma tendência entre essas duas variáveis o que demonstra que quanto maior o conhecimento em microbiologia, maior tende a ser a frequência de lavagem das mãos.

## Porém, essa relação não significa ao certo que o maior conhecimento influencie diretamente no aumento da frequência de lavagem, pois outros fatores podem influenciar esse comportamento.

## Em relação ao uso de álcool em gel, as respostas demonstraram um certo equilíbrio: 7 estudantes (36,8%) afirmaram utilizar álcool em gel, 7 (36,8%) afirmaram não utilizar e 5 (26,3%) afirmaram que utilizam às vezes. Ou seja, não houve unanimidade, indicando que o uso de álcool em gel é variável entre os estudantes e não correlacionado com conhecimento.

## Por fim, os resultados sugerem uma associação mais evidente entre o conhecimento em microbiologia e a frequência de lavagem das mãos do que em relação ao uso de álcool em gel. Os dados analisados permitem identificar essas tendências na amostra estudada, mas não que o conhecimento em microbiologia seja o responsável direto pelos hábitos de higiene dos estudantes.
