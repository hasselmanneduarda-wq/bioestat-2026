```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
```

```python
# Parâmetros do Exercício 1
mu1, sigma1 = 420, 55

# 1. Probabilidade de pesar menos de 350 g: P(X < 350)
p1 = stats.norm.cdf(350, loc=mu1, scale=sigma1)

# 2. Probabilidade de pesar mais de 500 g: P(X > 500)
p2 = 1 - stats.norm.cdf(500, loc=mu1, scale=sigma1)

# 3. Probabilidade de pesar entre 400 e 480 g: P(400 <= X <= 480)
p3 = stats.norm.cdf(480, loc=mu1, scale=sigma1) - stats.norm.cdf(
    400, loc=mu1, scale=sigma1
)

print("--- EXERCÍCIO 1 ---")
print(f"1. P(X < 350 g) = {p1:.4f} ({p1*100:.2f}%)")
print(f"2. P(X > 500 g) = {p2:.4f} ({p2*100:.2f}%)")
print(f"3. P(400 <= X <= 480 g) = {p3:.4f} ({p3*100:.2f}%)")
```

```python
# Curva da distribuição normal para os gráficos
x1 = np.linspace(mu1 - 4 * sigma1, mu1 + 4 * sigma1, 1000)
y1 = stats.norm.pdf(x1, loc=mu1, scale=sigma1)

fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)

# Gráfico 1: P(X < 350)
axes[0].plot(x1, y1, "k")
axes[0].fill_between(x1, y1, where=(x1 < 350), color="skyblue", alpha=0.6)
axes[0].set_title(f"P(X < 350 g) = {p1:.2%}")
axes[0].set_xlabel("Massa (g)")
axes[0].set_ylabel("Densidade de Probabilidade")

# Gráfico 2: P(X > 500)
axes[1].plot(x1, y1, "k")
axes[1].fill_between(x1, y1, where=(x1 > 500), color="salmon", alpha=0.6)
axes[1].set_title(f"P(X > 500 g) = {p2:.2%}")
axes[1].set_xlabel("Massa (g)")

# Gráfico 3: P(400 <= X <= 480)
axes[2].plot(x1, y1, "k")
axes[2].fill_between(
    x1, y1, where=((x1 >= 400) & (x1 <= 480)), color="lightgreen", alpha=0.6
)
axes[2].set_title(f"P(400 <= X <= 480 g) = {p3:.2%}")
axes[2].set_xlabel("Massa (g)")

plt.tight_layout()
plt.show()
```

```python
# Parâmetros do Exercício 2
mu2, sigma2 = 150, 10.5

# 1. Percentil 80 (P80)
p80 = stats.norm.ppf(0.80, loc=mu2, scale=sigma2)

# 2. Valor correspondente aos 10% superiores (Percentil 90 / P90)
p90 = stats.norm.ppf(0.90, loc=mu2, scale=sigma2)

# Percentil adicional (P50 - Mediana) para compor os 3 percentis solicitados no gráfico
p50 = stats.norm.ppf(0.50, loc=mu2, scale=sigma2)

print("--- EXERCÍCIO 2 ---")
print(f"1. Percentil 80: {p80:.2f} cm")
print(f"2. Limiar dos 10% maiores (Percentil 90): {p90:.2f} cm")
print(f"   (Mediana / Percentil 50: {p50:.2f} cm)")
```

```python
x2 = np.linspace(mu2 - 4 * sigma2, mu2 + 4 * sigma2, 1000)
y2 = stats.norm.pdf(x2, loc=mu2, scale=sigma2)

plt.figure(figsize=(9, 5))
plt.plot(x2, y2, "b-", lw=2, label="Distribuição de Comprimento")

# Indicadores verticais dos 3 percentis
plt.axvline(
    p50,
    color="gray",
    linestyle="--",
    lw=1.5,
    label=f"P50 (Mediana): {p50:.1f} cm",
)
plt.axvline(
    p80,
    color="orange",
    linestyle="--",
    lw=1.5,
    label=f"P80: {p80:.1f} cm",
)
plt.axvline(
    p90,
    color="red",
    linestyle="--",
    lw=1.5,
    label=f"P90 (Top 10%): {p90:.1f} cm",
)

plt.title("Percentis de Comprimento Corporal dos Mamíferos")
plt.xlabel("Comprimento (cm)")
plt.ylabel("Densidade")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

Exercicio 2 (interpretação)

A grande maioria da população ($80\%$) mede até $158,84\text{ cm}$, enquanto o valor de $163,46\text{ cm}$ (percentil 90) marca o limite a partir do qual estão os $10\%$ maiores indivíduos. Na prática, estes indivíduos de maior porte costumam ter vantagens adaptativas, como maior facilidade para dominar territórios, obter parceiros reprodutivos e competir por recursos.

```python
# Parâmetros do Exercício 3
mu3, sigma3 = 72, 8

# 1. P(X <= 60)
p_a = stats.norm.cdf(60, loc=mu3, scale=sigma3)

# 2. P(X >= 85)
p_b = 1 - stats.norm.cdf(85, loc=mu3, scale=sigma3)

# 3. P(65 <= X <= 80)
p_c = stats.norm.cdf(80, loc=mu3, scale=sigma3) - stats.norm.cdf(
    65, loc=mu3, scale=sigma3
)

# 4. Percentis 90, 95 e 99
p90_aves = stats.norm.ppf(0.90, loc=mu3, scale=sigma3)
p95_aves = stats.norm.ppf(0.95, loc=mu3, scale=sigma3)
p99_aves = stats.norm.ppf(0.99, loc=mu3, scale=sigma3)

print("--- EXERCÍCIO 3 ---")
print(f"1. P(X <= 60 g) = {p_a:.4f} ({p_a*100:.2f}%)")
print(f"2. P(X >= 85 g) = {p_b:.4f} ({p_b*100:.2f}%)")
print(f"3. P(65 <= X <= 80 g) = {p_c:.4f} ({p_c*100:.2f}%)")
print(f"4. Percentil 90 = {p90_aves:.2f} g")
print(f"   Percentil 95 = {p95_aves:.2f} g")
print(f"   Percentil 99 = {p99_aves:.2f} g")
```

```python
x3 = np.linspace(mu3 - 4 * sigma3, mu3 + 4 * sigma3, 1000)
y3 = stats.norm.pdf(x3, loc=mu3, scale=sigma3)

fig, ax = plt.subplots(figsize=(11, 6))

ax.plot(x3, y3, "k-", lw=2, label="Distribuição da Massa Corporal")

# Regiões de Probabilidade
ax.fill_between(
    x3, y3, where=(x3 <= 60), color="blue", alpha=0.3, label="P(X <= 60 g)"
)
ax.fill_between(
    x3, y3, where=(x3 >= 85), color="red", alpha=0.3, label="P(X >= 85 g)"
)
ax.fill_between(
    x3,
    y3,
    where=((x3 >= 65) & (x3 <= 80)),
    color="green",
    alpha=0.3,
    label="P(65 <= X <= 80 g)",
)

# Linhas de Percentis
ax.axvline(
    p90_aves,
    color="purple",
    linestyle="--",
    lw=1.5,
    label=f"P90 ({p90_aves:.1f} g)",
)
ax.axvline(
    p95_aves,
    color="orange",
    linestyle="--",
    lw=1.5,
    label=f"P95 ({p95_aves:.1f} g)",
)
ax.axvline(
    p99_aves,
    color="brown",
    linestyle="--",
    lw=1.5,
    label=f"P99 ({p99_aves:.1f} g)",
)

ax.set_title("Distribuição Normal de Massa Corporal de Aves")
ax.set_xlabel("Massa Corporal (g)")
ax.set_ylabel("Densidade de Probabilidade")
ax.legend(loc="upper left")
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Exercicio 3 (interpretação)

A população apresenta forte seleção estabilizadora: cerca de $65\%$ das aves mantêm uma massa intermediária (entre $65\text{ g}$ e $80\text{ g}$), ideal para o equilíbrio diário. Apenas $6,7\%$ pesam $60\text{ g}$ ou menos (indivíduos mais frágeis ou subnutridos) e cerca de $5\%$ superam $85\text{ g}$ (percentil 95 em $85,16\text{ g}$). Ter mais massa ajuda em reservas para migrações, mas exige mais gasto energético no voo e reduz a agilidade contra predadores.
