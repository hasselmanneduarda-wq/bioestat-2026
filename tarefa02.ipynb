```python
import pandas as pd
import matplotlib.pyplot as plt

dados = pd.read_excel("semana02.xlsx")

dados.columns = dados.columns.str.strip()

dados
```

```python
contagem = dados["Cultivar"].value_counts()

cores = ["#6C8CD5", "#E889B5", "#9B7ED9"]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    contagem.index,
    contagem.values,
    color=cores,
    edgecolor="white",
    linewidth=1.5
)

plt.title(
    "Frequência das Cultivares de Soja",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel("Cultivar", fontsize=12)
plt.ylabel("Frequência", fontsize=12)

for barra in barras:
    altura = barra.get_height()
    plt.text(
        barra.get_x() + barra.get_width()/2,
        altura + 0.2,
        f"{int(altura)}",
        ha="center",
        fontsize=11
    )

plt.show()
```

```python
media_vagens = dados.groupby("Cultivar ")["N de vagens"].mean()

cores = ["#6C8CD5", "#E889B5", "#9B7ED9"]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    media_vagens.index,
    media_vagens.values,
    color=cores,
    edgecolor="white",
    linewidth=1.5
)

plt.title(
    "Nº Médio de Vagens por Cultivar de Soja",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel("Cultivar ", fontsize=12)
plt.ylabel("Nº médio de vagens", fontsize=12)

for barra in barras:
    altura = barra.get_height()
    plt.text(
        barra.get_x() + barra.get_width()/2,
        altura + 0.5,
        f"{altura:.1f}",
        ha="center",
        fontsize=11
    )

plt.show()
```

```python
prod_A = dados.loc[dados["Cultivar "] == "A", "Produtividade (kg/ha)"]
prod_B = dados.loc[dados["Cultivar "] == "B", "Produtividade (kg/ha)"]
prod_C = dados.loc[dados["Cultivar "] == "C", "Produtividade (kg/ha)"]

cores = ["#6C8CD5", "#E889B5", "#9B7ED9"]

plt.figure(figsize=(8, 5))

box = plt.boxplot(
    [prod_A, prod_B, prod_C],
    tick_labels=["A", "B", "C"],
    patch_artist=True
)

for caixa, cor in zip(box["boxes"], cores):
    caixa.set_facecolor(cor)

plt.title(
    "Produtividade por Cultivar de Soja",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel("Cultivar ", fontsize=12)
plt.ylabel("Produtividade (kg/ha)", fontsize=12)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    dados["N de vagens"],
    bins=8,
    color="#9B7ED9",
    edgecolor="white",
    linewidth=1.5
)

plt.title(
    "Nº de Vagens em Plantas de Soja",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel("Nº de vagens", fontsize=12)
plt.ylabel("Frequência", fontsize=12)

plt.show()
```
